```python
import pandas as pd
import os
from datetime import datetime

# ==========================================
# 1. CONFIGURACIÓN EDITABLE
# ==========================================
METAS = {
    "31-60": 2208249.80,   #
    "61-90": 4184439.54,   #
    "91-120": 1010891.51,  #
    "TOTAL": 7403580.85    #
}

# Configuración manual de Reposesiones[cite: 1]
META_REPOS = 13
REPOS_LOGRADAS = 5  # <--- Actualizar manualmente conforme recuperen unidades

RUTA_PAGOS = r'C:\Users\SISTEMAS\Documents\MAQUINA DIEGO\CARTERAS\GMF\PAGO IMPORTACION\2026\04 ABRIL'
DIAS_LABORADOS_MES = 24
DIAS_TRANSCURRIDOS = 10 
TOTAL_ASESORES = 15

# ==========================================
# 2. CONSULTAS SQL (PARA TU REFERENCIA)
# ==========================================
QUERY_PAGOS = """
SELECT CAST([numcredito] AS VARCHAR(25)) AS numcredito, [monto_pago], [fecha_pago], [aging]
FROM [Asignacion].[dbo].[CONVENIOS_FULL]
WHERE id_Cliente = 'GMF' AND MONTH([fecha_pago]) = MONTH(GETDATE())
"""

QUERY_RPC = """
SELECT a.ID_CUENTA, a.ESTATUS, b.aging
FROM [METAGABSSA_2025].[dbo].[historico_resumen_gestion] a 
LEFT JOIN [BAHIA].[dbo].[MACRO_GMF] b ON a.id_cuenta = b.id_cuenta
WHERE a.Id_Cliente IN ('GMF_PRELEGAL','GMF') AND a.Accion <> 'Gestor Visita'
"""

# ==========================================
# 3. PROCESAMIENTO DE DATOS
# ==========================================
def obtener_bucket(mora):
    try:
        val = int(mora)
        if 31 <= val <= 60: return "31-60"
        if 61 <= val <= 90: return "61-90"
        if 91 <= val <= 120: return "91-120"
    except:
        pass
    return "Otros"

def generar_reporte():
    archivos = [os.path.join(RUTA_PAGOS, f) for f in os.listdir(RUTA_PAGOS) if f.endswith('.xlsx')]
    if not archivos:
        print("No se encontraron archivos Excel en la ruta especificada.")
        return

    dfs = []
    for f in archivos:
        temp = pd.read_excel(f)
        # Normalizamos nombres: minúsculas y sin espacios locos
        temp.columns = temp.columns.str.strip().str.lower()
        dfs.append(temp)
    
    df = pd.concat(dfs, ignore_index=True)

    # MAPEÓN DE COLUMNAS SEGÚN TU ERROR[cite: 1]
    col_cta = 'contrato' if 'contrato' in df.columns else 'numcredito'
    col_monto = 'monto de pago' if 'monto de pago' in df.columns else 'monto_pago'
    col_fecha = 'fecha de pago' if 'fecha de pago' in df.columns else 'fecha_pago'
    col_mora = 'días de mora' if 'días de mora' in df.columns else 'aging'

    # Limpieza de Notación Científica y Duplicados[cite: 1]
    df[col_cta] = df[col_cta].astype(str).str.split('.').str[0]
    df['fecha_dt'] = pd.to_datetime(df[col_fecha]).dt.date
    df = df.drop_duplicates(subset=[col_cta, 'fecha_dt'])

    # Clasificación por Bucket
    df['bucket_calc'] = df[col_mora].apply(obtener_bucket)

    # --- CÁLCULOS ---
    total_recup = df[col_monto].sum()
    dias_rest = DIAS_LABORADOS_MES - DIAS_TRANSCURRIDOS
    rpc_ejemplo = 11.2 # Este valor lo puedes automatizar con tu query de gestión[cite: 1]

    # Filas de la Tabla HTML
    filas_html = ""
    for b_name, b_meta in METAS.items():
        if b_name == "TOTAL": continue
        recup = df[df['bucket_calc'] == b_name][col_monto].sum()
        avance = (recup / b_meta * 100) if b_meta > 0 else 0
        falta = b_meta - recup
        diario_asesor = (falta / dias_rest / TOTAL_ASESORES) if dias_rest > 0 else 0
        
        color_class = "verde" if avance >= (DIAS_TRANSCURRIDOS/DIAS_LABORADOS_MES*100) else "rojo"
        
        filas_html += f"""
        <tr>
            <td>Bucket {b_name}</td>
            <td>${b_meta:,.2f}</td>
            <td class="{color_class}">${recup:,.2f}</td>
            <td>{avance:.1f}%</td>
            <td><b>${diario_asesor:,.2f}</b></td>
        </tr>
        """

    # --- DISEÑO CSS Y HTML ---
    html_final = f"""
    <html>
    <head>
        <style>
            body {{ font-family: 'Segoe UI', sans-serif; margin: 30px; background-color: #f8f9fa; }}
            .header {{ background: #003566; color: white; padding: 25px; border-radius: 12px; text-align: center; box-shadow: 0 4px 10px rgba(0,0,0,0.2); }}
            .container {{ display: flex; gap: 20px; margin: 25px 0; }}
            .card {{ background: white; padding: 20px; border-radius: 12px; flex: 1; box-shadow: 0 4px 6px rgba(0,0,0,0.05); text-align: center; border-top: 5px solid #003566; }}
            .card h2 {{ margin: 0; color: #6c757d; font-size: 15px; text-transform: uppercase; }}
            .card p {{ font-size: 26px; font-weight: bold; color: #003566; margin: 10px 0; }}
            table {{ width: 100%; border-collapse: collapse; background: white; border-radius: 12px; overflow: hidden; box-shadow: 0 4px 6px rgba(0,0,0,0.05); }}
            th {{ background: #003566; color: white; padding: 18px; font-size: 14px; }}
            td {{ padding: 15px; border-bottom: 1px solid #dee2e6; text-align: center; font-size: 15px; }}
            .verde {{ color: #2d6a4f; font-weight: bold; }}
            .rojo {{ color: #a4133c; font-weight: bold; }}
            .progress-bg {{ background: #e9ecef; height: 12px; border-radius: 6px; margin-top: 10px; overflow: hidden; }}
            .progress-fill {{ background: #ffc300; height: 100%; width: {(REPOS_LOGRADAS/META_REPOS*100):.1f}%; }}
        </style>
    </head>
    <body>
        <div class="header">
            <h1>GABSSA: Seguimiento de Recuperación GMF</h1>
            <p>Estado al día {DIAS_TRANSCURRIDOS} de {DIAS_LABORADOS_MES} (Abril 2026)</p>
        </div>
        <div class="container">
            <div class="card"><h2>RPC Diario</h2><p>{rpc_ejemplo}%</p><small>Meta: 10-12%[cite: 1]</small></div>
            <div class="card"><h2>Recuperado Total</h2><p>${total_recup:,.2f}</p><span>{(total_recup/METAS['TOTAL']*100):.1f}% de la Meta</span></div>
            <div class="card">
                <h2>Unidades Repo</h2>
                <p>{REPOS_LOGRADAS} / {META_REPOS}</p>
                <div class="progress-bg"><div class="progress-fill"></div></div>
            </div>
        </div>
        <table>
            <thead>
                <tr>
                    <th>Segmento</th>
                    <th>Meta del Mes</th>
                    <th>Recuperado Real</th>
                    <th>% Avance</th>
                    <th>Meta Diaria por Asesor</th>
                </tr>
            </thead>
            <tbody>{filas_html}</tbody>
        </table>
    </body>
    </html>
    """
    
    with open("Dash_Recuperacion_GMF.html", "w", encoding="utf-8") as f:
        f.write(html_final)
    print(f"¡Éxito! Reporte generado como 'Dash_Recuperacion_GMF.html'. Columna usada: {col_cta}")

if __name__ == "__main__":
    generar_reporte()
```
